```python
W_0 = np.array(W_0, dtype=float)
b_0 = np.array(b_0, dtype=float)

W_1 = np.array(W_1, dtype=float)
b_1 = np.array(b_1, dtype=float)
```

```python
import numpy as np

def relu(Z):
    return np.maximum(0, Z)

def relu_derivative(Z):
    return (Z > 0).astype(float)

def softmax(Z):
    Z = Z - np.max(Z)
    exp_Z = np.exp(Z)
    return exp_Z / np.sum(exp_Z)


def train_one_image(x, label, learning_rate=0.01):
    global W_0, b_0, W_1, b_1

    # STEP 1: Forward propagation
    Z_1 = np.dot(W_0, x) + b_0
    A_1 = relu(Z_1)

    Z_2 = np.dot(W_1, A_1) + b_1
    P = softmax(Z_2)

    # STEP 2: Correct answer (one-hot)
    Y = np.zeros(10)
    Y[int(label)] = 1.0

    # STEP 3: Calculate loss
    loss = -np.log(np.clip(P[int(label)], 1e-12, 1.0))

    # STEP 4: Output layer gradients
    dZ_2 = P - Y

    dW_1 = np.outer(dZ_2, A_1)
    db_1 = dZ_2

    # STEP 5: Hidden layer gradients
    # Use the OLD W_1 before updating it
    dZ_1 = (W_1.T @ dZ_2) * relu_derivative(Z_1)

    dW_0 = np.outer(dZ_1, x)
    db_0 = dZ_1

    # STEP 6: Gradient descent
    W_1 -= learning_rate * dW_1
    b_1 -= learning_rate * db_1

    W_0 -= learning_rate * dW_0
    b_0 -= learning_rate * db_0

    return loss, int(np.argmax(P))
```

```python
# Convert the existing inputs to arrays
X_train = np.asarray(X, dtype=float)

# This assumes label is the first column
y_train = np.asarray(data[:, 0], dtype=int)

print("Images:", X_train.shape)
print("Labels:", y_train.shape)
```

```python
epochs = 10
learning_rate = 0.01

for epoch in range(epochs):

    total_loss = 0.0
    correct = 0

    # Shuffle training examples every epoch
    indices = np.random.permutation(len(X_train))

    for i in indices:

        x = X_train[i]
        label = y_train[i]

        loss, prediction = train_one_image(
            x,
            label,
            learning_rate
        )

        total_loss += loss

        if prediction == label:
            correct += 1

    avg_loss = total_loss / len(X_train)
    accuracy = correct / len(X_train) * 100

    print(
        f"Epoch {epoch + 1}/{epochs} | "
        f"Loss: {avg_loss:.4f} | "
        f"Accuracy: {accuracy:.2f}%"
    )
```

```python
def predict(x):
    Z_1 = W_0 @ x + b_0
    A_1 = relu(Z_1)

    Z_2 = W_1 @ A_1 + b_1
    P = softmax(Z_2)

    return int(np.argmax(P)), P


predicted_digit, probabilities = predict(X_train[0])

print("Actual:", y_train[0])
print("Predicted:", predicted_digit)
print("Probabilities:", probabilities)
```
