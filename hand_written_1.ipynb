```python
X = []

for i in range(len(data)):
  X.append(data[i][1:]/255)
```

```python
len(X[0])
```

```python

def he_distribution(n_in):
    std = np.sqrt(2 / n_in)
    return np.random.normal(0, std, size=n_in)

W_0 = he_distribution(len(X[0]))
```

```python
b0 = 0.0

Z_1 = np.dot(W_0, X[0]) + b0
A_1 = np.maximum(0, Z_1)
```

```python
print(Z_1)
```

```python

W_0 = []
b_0 = []
Z_1 = []
A_1 = []

n_in = len(X[0])

for i in range(10):
    weights = np.random.normal(
        0,
        np.sqrt(2 / n_in),
        size=n_in
    )

    W_0.append(weights)
    b_0.append(0.0)

    z = np.dot(weights, X[0]) + b_0[i]
    Z_1.append(z)

    a = max(0, z)
    A_1.append(a)
```

```python
W_1 = []
b_1 = []

n_in = len(A_1)  # 10

for i in range(10):
    weights = np.random.normal(
        0,
        np.sqrt(1 / n_in),
        size=n_in
    )

    W_1.append(weights)
    b_1.append(0.0)
```

```python
print(np.shape(W_1))  # (10, 10)
print(np.shape(b_1))  # (10,)
```

```python
Z_2 = np.dot(W_1, A_1) + np.array(b_1)

print("Output scores:", Z_2)
```

```python
def softmax(Z):
    exp_Z = np.exp(Z - np.max(Z))
    return exp_Z / np.sum(exp_Z)

P = softmax(Z_2)

print("Probabilities:", P)
print("Total:", np.sum(P))
```

```python
predicted_digit = np.argmax(P)

print("Predicted digit:", predicted_digit)
```
